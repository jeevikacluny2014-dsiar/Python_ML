```python
for i in df['duration'].values:
  print(i)
  if(('h' in i) and ('m' in i))==True:
    print('if')
    print(int(re.findall(r'(\d+)h',i)[0])*60 + int(re.findall(r'(\d+)m',i)[0]))
  elif('h'in i)==True:
    print('elif-1')
    print(int(re.findall(r'(\d+)h',i)[0])*60)
  elif('m'in i)==True:
    print('elif-2')
    print(int(re.findall(r'(\d+)m',i)[0]))
```

```python
'm' in '2h 22m'
```

```python
list_dur_min=[]
for i in df['duration'].values:
  print(i)
  if(('h' in i) and ('m' in i))==True:
    print('if')
    print(int(re.findall(r'(\d+)h',i)[0])*60 + int(re.findall(r'(\d+)m',i)[0]))
    list_dur_min.append(int(re.findall(r'(\d+)h',i)[0])*60 + int(re.findall(r'(\d+)m',i)[0]))
  elif('h'in i)==True:
    print('elif-1')
    print(int(re.findall(r'(\d+)h',i)[0])*60)
    list_dur_min.append(int(re.findall(r'(\d+)h',i)[0])*60)
  elif('m'in i)==True:
    print('elif-2')
    print(int(re.findall(r'(\d+)m',i)[0]))
    list_dur_min.append(int(re.findall(r'(\d+)m',i)[0]))
```
